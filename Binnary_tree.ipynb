```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

```python
# Creating nodes
root = Node(1)
root.left = Node(2)
root.right = Node(3)

root.left.left = Node(4)
root.left.right = Node(5)
root.left.right.left = Node(8)
```

```python
print(root.left.right.left.data)
```

- print the data in PREORDER 

```python
def preOrderTraversal(node):
    if node is None:
        return
    print(node.data, end=", ")
    preOrderTraversal(node.left)
    preOrderTraversal(node.right)
```

```python
preOrderTraversal(root)
```

```python
def Postorder(node):
    if node is None:
        return
    Postorder(node.left)
    Postorder(node.right)
    print(node.data,end= ' ')
```

```python
Postorder(root)
```

```python
def inorder(node):
    if node is None:
        return
    inorder(node.left)
    print(node.data,end=' ')
    inorder(node.right)
```

```python
inorder(root)
```

- Traversing binnary tree (PRE ORDER):

```python

def Pre_order(root):
    if  root is None:
        return

    # Step 1: Initialize stack and pointer
    stack = [None]
    ptr = root

    # Step 2: Repeat while ptr is not Null
    while ptr is not None:
        # a) Apply process to info[ptr]
        print(ptr.data, end=" ")

        # b) If right child exists, save it to the stack
        if ptr.right is not None:
            stack.append(ptr.right)

        # c) If left child exists, move to it
        if ptr.left is not None:
            ptr = ptr.left
        else:
            # d) Otherwise, pop the next available node from the stack
            ptr = stack.pop()

```

```python
Pre_order(root)
```

```python
def post_order(root):
    stack = [None]
    ptr = root
    
    while True:
        # --- GOING DOWN ---
        while ptr is not None:
            stack.append(ptr)           # Step 3: Save parent
            if ptr.right is not None:
                # Step 4: Save Right with "-ve sign" 
                # We use a list [ptr.right] to represent "Negative"
                stack.append(ptr.right) 
            ptr = ptr.left              # Step 5: move left
            
        # --- COMING UP ---
        ptr = stack.pop()               # Step 6: Take value from stack
        
        # Step 7: While it's "Positive" (a normal node), process it
        while ptr is not None and not isinstance(ptr, list):
            print(ptr.data, end=" ")    # Apply process
            ptr = stack.pop()           # Take next value
            
        # Step 8: If it's "Negative" (the list we made)
        if ptr is not None and isinstance(ptr, list):
            ptr = ptr[1]                # Step 8a: Make it positive (get the node)
            continue                    # Step 8b: Goto step 2 (restart loop)

        if ptr is None: break           # Step 9: Exit
```

```python
post_order(root)
```
