```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
n1 = Node(32)
n2 = Node(45)
n1.next = n2 
n3 = Node(30)
n2.next = n3
head1 = n1
```

```python
n4 = Node(2)
n5 = Node(95)
n4.next = n5 
n6 = Node(80)
n5.next = n6
head2 = n4
```

```python
result = 0
while head1 is not None:
    print(head1.data, end=" ")
    result += head1.data 
    head1 = head1.next 
print("\n")

print("Sum of all nodes in linked list is:", result)
```

```python
result1 = 0
while head2 is not None:
    print(head2.data, end=" ")
    result1 += head2.data 
    head2 = head2.next 
print("\n")

print("Sum of all nodes in linked list is:", result1)
```

```python
final_list = result + result1
print("Final list:", final_list)
```

```python
while final_list:
    num = final_list % 10
    node7 = Node(num)
    final_list = final_list // 10
    num = num % 10
    node8 = Node(num)
    node7.next = node8
    final_list = final_list // 10
    num = num % 10
    node9 = Node(num)
    node8.next = node9
```

```python
head3 = node7
while head3 is not None:
    print(head3.data, end=" ")
    head3 = head3.next
```
