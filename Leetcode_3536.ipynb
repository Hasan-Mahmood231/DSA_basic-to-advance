```python
class Node:
    def __init__(self, data):
        self.data = data 
        self.next = None
```

```python
n1 = Node(2)
n2 = Node(4)
n3 = Node(6)
n1.next = n2
n2.next = n3
n3.next = None

# now print all these 
first = 0
second = 0
while n1 :
    if n1.data > first:
        second = first
        first = n1.data
    elif n1.data > second and n1.data != first:
        second = n1.data
    n1 = n1.next


print(first * second)
```

```python
prev = 0
nxt = 0
for i in range(1, 11):
    if i > prev:
        nxt = prev
        prev = i
    elif i > nxt:
        nxt = i
        

print("Largest:", prev)
print("Second Largest:", nxt)
```
